## Parte 2: Construcción de un índice invertido

### Actividad
1. Escribe un programa que:
   - Recorra todos los documentos.
   - Construya un **índice invertido**, es decir, un diccionario donde:
     - Cada palabra clave apunta a una lista de documentos donde aparece.

2. Escribe una nueva función de búsqueda que:
   - Consulte directamente el índice para encontrar los documentos relevantes.
   - Sea mucho más rápida que la búsqueda lineal.

```python
from collections import defaultdict

documentos = ["/content/01_corpus_turismo.txt","/content/01_corpus_turismo_500.txt"]

def construir_indice(documentos):
    diccionario = {}
    indice_invertido = defaultdict(lambda: defaultdict(int))  # uso de defaultdict para palabras
    for documento in documentos:
        with open(documento, 'r', encoding='utf-8') as f:
            texto = f.read()
            palabras = texto.lower().split()
            for palabra in palabras:
                palabra = palabra.strip(".,;:¡!¿?()[]{}\"")

                # Agregamos la palabra en el diccionario
                if palabra not in diccionario:
                    diccionario[palabra] = 1

                # Update el contador de la palabra en el indice invertido
                indice_invertido[palabra][documento] += 1  # Incrementa el contador por cada palabra encontrada en el documento

    return diccionario, indice_invertido

# Build el diccionario y el indice
diccionario, indice = construir_indice(documentos)

# Function para buscar la palabra
def buscar_en_indice(palabra, indice):
    palabra = palabra.lower().strip(".,;:¡!¿?()[]{}\"")
    if palabra in indice:
        for documento, count in indice[palabra].items():
            print(f"La palabra '{palabra}' aparece en el documento '{documento}', {count} veces.")
    else:
        print(f"La palabra '{palabra}' no se encontró en ningún documento.")


palabra_buscada = input("Introduce la palabra a buscar usando el índice invertido: ")
buscar_en_indice(palabra_buscada, indice)
```

## Parte 3: Evaluación de tiempos de búsqueda
### Actividad

1. Realiza la búsqueda de varias palabras usando:
      -  Corpus pequeño: 16 documentos (turismo en Ecuador).

2. Mide el tiempo de ejecución:

### Ejemplo de palabras para buscar

- playas

```python

import time
from collections import defaultdict

documentos_pequenos = ["/content/01_corpus_turismo.txt"]
documentos_grandes = ["/content/01_corpus_turismo_500.txt"]
#palabras_clave = ["quito", "montañita", "feriado", "playas", "aventura", "galápagos"]
palabras_clave = input("Introduce las palabras a buscar separadas por comas: ").lower().split(",")

# Función de búsqueda lineal
def busqueda_lineal(palabra, documentos):
    inicio = time.time()
    for documento in documentos:
        with open(documento, 'r', encoding='utf-8') as f:
            for linea in f:
                if palabra in linea.lower():
                    pass  # Simulamos la acción de encontrar la palabra
    fin = time.time()
    return fin - inicio

def construir_indice(documentos):
    indice_invertido = defaultdict(list)
    for documento in documentos:
        with open(documento, 'r', encoding='utf-8') as f:
            for linea in f:
                for palabra in linea.lower().split():
                    palabra = palabra.strip(".,;:¡!¿?()[]{}\"")
                    indice_invertido[palabra].append(documento)
    return indice_invertido

def buscar_en_indice(palabra, indice):
    inicio = time.time()
    if palabra in indice:
        pass  # Simulamos la acción de encontrar la palabra en el índice
    fin = time.time()
    return fin - inicio

# Resultados
resultados = []

for palabra in palabras_clave:
    tiempo_lineal_pequeno = busqueda_lineal(palabra, documentos_pequenos)
    indice_pequeno = construir_indice(documentos_pequenos)
    tiempo_indice_pequeno = buscar_en_indice(palabra, indice_pequeno)

    tiempo_lineal_grande = busqueda_lineal(palabra, documentos_grandes)
    indice_grande = construir_indice(documentos_grandes)
    tiempo_indice_grande = buscar_en_indice(palabra, indice_grande)

    resultados.append([palabra, tiempo_lineal_pequeno, tiempo_indice_pequeno, tiempo_lineal_grande, tiempo_indice_grande])

# Imprimir tabla de resultados
print("Palabra | Tiempo Lineal (Pequeño) | Tiempo Índice (Pequeño) | Tiempo Lineal (Grande) | Tiempo Índice (Grande)")
print("-" * 100)
for resultado in resultados:
    print(f"{resultado[0]} | {resultado[1]:.6f} s | {resultado[2]:.6f} s | {resultado[3]:.6f} s | {resultado[4]:.6f} s")
```

2. Permite consultas de múltiples términos (ejemplo: buscar documentos que contengan "playa" y "turismo").
3. Calcula el _speedup_

```python

from collections import defaultdict
import time


def construir_indice(documentos):
    indice_invertido = defaultdict(list)
    for documento in documentos:
        with open(documento, 'r', encoding='utf-8') as f:
            for linea in f:
                for palabra in linea.lower().split():  # Convertir a minúsculas aquí
                    palabra = palabra.strip(".,;:¡!¿?()[]{}\"")
                    indice_invertido[palabra].append(documento)
    return indice_invertido

def buscar_en_indice(palabras, indice):
    inicio = time.time()
    resultados = set()  # Usar un conjunto para evitar duplicados
    for palabra in palabras:
        palabra = palabra.lower().strip(".,;:¡!¿?()[]{}\"") #Convertir a minúsculas
        if palabra in indice:
            if not resultados:  # Inicializar resultados con el primer conjunto de documentos
                resultados = set(indice[palabra])
            else:
                resultados = resultados.intersection(set(indice[palabra]))  # Intersección de conjuntos
    fin = time.time()
    return fin - inicio, resultados

# Obtener la entrada del usuario para las palabras clave
palabras_clave = input("Introduce las palabras a buscar separadas por comas: ").lower().split(",")

# Resultados
resultados = []

for palabra in palabras_clave:
    tiempo_lineal_pequeno = busqueda_lineal(palabra, documentos_pequenos)
    indice_pequeno = construir_indice(documentos_pequenos)
    tiempo_indice_pequeno, _ = buscar_en_indice([palabra], indice_pequeno) # Pasar una lista con una sola palabra clave

    tiempo_lineal_grande = busqueda_lineal(palabra, documentos_grandes)
    indice_grande = construir_indice(documentos_grandes)
    tiempo_indice_grande, _ = buscar_en_indice([palabra], indice_grande) # Pasar una lista con una sola palabra clave


    resultados.append([palabra, tiempo_lineal_pequeno, tiempo_indice_pequeno, tiempo_lineal_grande, tiempo_indice_grande])

# Imprimir tabla de resultados
print("Palabra | Tiempo Lineal (Pequeño) | Tiempo Índice (Pequeño) | Tiempo Lineal (Grande) | Tiempo Índice (Grande)")
print("-" * 100)
for resultado in resultados:
    print(f"{resultado[0]} | {resultado[1]:.6f} s | {resultado[2]:.6f} s | {resultado[3]:.6f} s | {resultado[4]:.6f} s")

# Calcular speedup
for resultado in resultados:
  speedup_pequeno = resultado[1] / resultado[2] if resultado[2] !=0 else "Infinito"
  speedup_grande = resultado[3] / resultado[4] if resultado[4] !=0 else "Infinito"
  print(f"Speedup pequeño para {resultado[0]}: {speedup_pequeno}")
  print(f"Speedup grande para {resultado[0]}: {speedup_grande}")
```
