*   Cuantos terminos hay en el corpus




```python

def corpus_terms(corpus):
    term_list = []
    for document_path in corpus:
        with open(document_path, 'r') as file:
            document_content = file.read()
            terms = document_content.split()
            term_list.extend(terms)
    return len(term_list)

document = ["/home/01_corpus_turismo.txt"]
total_terms = corpus_terms(document)
print(total_terms)
```

* Eliminar tildes, mayusculas, y que enliste en una tabla los terminos mas repetidos


```python
import unicodedata
import pandas as pd
import re

# Lista de documentos
documentos = ["/home/01_corpus_turismo.txt", "/home/01_corpus_turismo_500.txt"]

# Función para limpiar el texto
def limpiar_texto(texto):
    texto = texto.lower()
    texto = unicodedata.normalize('NFD', texto)
    texto = ''.join(c for c in texto if unicodedata.category(c) != 'Mn')
    texto = re.sub(r'[^\w\s]', '', texto)
    return texto

# Contar palabras por documento
palabras_por_doc = {}
todos_los_terminos = set()

for i, ruta in enumerate(documentos):
    nombre_doc = f"doc{i+1}"
    with open(ruta, 'r') as archivo:
        texto = archivo.read()
    texto_limpio = limpiar_texto(texto)
    palabras = texto_limpio.split()
    palabras_por_doc[nombre_doc] = palabras
    todos_los_terminos.update(palabras)

terminos_ordenados = sorted(todos_los_terminos)

# Crear DataFrame de frecuencias (conteo)
frecuencias = pd.DataFrame(index=terminos_ordenados)

for doc, palabras in palabras_por_doc.items():
    conteo = {t: palabras.count(t) for t in terminos_ordenados}
    frecuencias[doc] = pd.Series(conteo)

frecuencias.index.name = 'Términos'
print(frecuencias)
```
