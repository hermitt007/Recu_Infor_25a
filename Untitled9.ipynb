
# **Parte 2: Obtener los datos deseados**
- Buscar dentro del contenido HTML y extraer la información.

```python
title = soup.find("meta", {"property": "og:title"})["content"]
print("Titulo: ",title)
```

```python
ingredients_section = soup.find_all("li", class_='mm-recipes-structured-ingredients__list-item')
print("Ingredientes: ")
for ingredient in ingredients_section:
    print(ingredient.text.strip())
```

```python
details_divs = soup.find_all("div", class_="mm-recipes-details__value")
details_text = None

if len(details_divs) > 2:
    details_text = details_divs[2].get_text(strip=True)

print("Tiempo total: ",details_text)
```

```python
rating_section = soup.find("div", id='mm-recipes-review-bar__rating_1-0')
for value in rating_section:
    print("Rating: ", value)
```

```python
# Extraer pasos de preparación sin créditos
preparation_steps = []

step_items = soup.select("li.mntl-sc-block-group--LI")

for i, item in enumerate(step_items, 1):
    text = item.get_text(" ", strip=True)
    if text and len(text) > 20:
        # Eliminar texto no deseado si está al final
        if text.endswith("Dotdash Meredith Food Studios"):
            text = text.replace("Dotdash Meredith Food Studios", "").strip()
        preparation_steps.append(f"step {i}: {text}")

# Imprimir los pasos limpios
for step in preparation_steps:
    print(step)
```

```python
recipes_links = soup.find_all("a")
recipe_urls = []
for link in recipes_links:
    href = link.get('href')
    if "recipe" in href:
        recipe_urls.append(href)
print ("Linked recipes:")
for url in recipe_urls:
    print(url)
print(len(recipe_urls))
```

```python
import requests
import time
from bs4 import BeautifulSoup

recipe_data = []

headers = {"User-Agent": "Mozilla/5.0"}

for url in recipe_urls:
    try:
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.text, "html.parser")

        title = soup.find("meta", {"property": "og:title"})
        title = title["content"] if title else "Título no encontrado"

        summary = next(
            (p.get_text(strip=True) for p in soup.find_all("p") if len(p.get_text(strip=True)) > 100),
            "Resumen no encontrado"
        )

        ingredients = " | ".join([
            ing.text.strip()
            for ing in soup.find_all("li", class_="mm-recipes-structured-ingredients__list-item")
        ]) or "Ingredientes no encontrados"

        steps = " ".join([
            f"step {i+1}: {item.get_text(' ', strip=True)}"
            for i, item in enumerate(soup.select("li.mntl-sc-block-group--LI"))
        ]) or "Pasos no encontrados"

        total_time = next(
            (value.text.strip() for label, value in zip(
                soup.find_all("div", class_="mm-recipes-details__label"),
                soup.find_all("div", class_="mm-recipes-details__value"))
             if "Total Time" in label.text),
            "Tiempo no encontrado"
        )

        servings = next(
            (value.text.strip() for label, value in zip(
                soup.find_all("div", class_="mm-recipes-details__label"),
                soup.find_all("div", class_="mm-recipes-details__value"))
             if "Servings" in label.text),
            "Porciones no encontradas"
        )

        rating_tag = soup.find("div", class_="comp mm-recipes-review-bar__rating mntl-text-block text-label-300")
        rating = rating_tag.text.strip() if rating_tag else "Rating no encontrado"

        nutrition = " | ".join([
            f"{row.find_all('td')[1].get_text(strip=True)}: {row.find_all('td')[0].get_text(strip=True)}"
            for row in soup.select("tr.mm-recipes-nutrition-facts-summary__table-row")
            if len(row.find_all('td')) >= 2
        ]) or "Nutrición no encontrada"

        # Agregar a la lista
        recipe_data.append({
            "url": url,
            "title": title,
            "summary": summary,
            "ingredients": ingredients_section,
            "steps": preparation_steps,
            "total_time": total_time,
            "servings": servings,
            "rating": rating
        })

        print(f"✅ Receta extraída: {title}")

    except Exception as e:
        print(f"❌ Error en {url}: {e}")

    time.sleep(1)  # evitar sobrecarga

# Mostrar un ejemplo
print("\nEjemplo de receta extraída:")
print(recipe_data[0])
```


#**Parte 4: Hacer RAG con las recetas obtenidas**
- Una vez que se ha construido el corpus, implementar y desplegar RAG para realizar búsquedas en el corpus

```python
cleaned_recipe_data = []

for recipe in recipe_data:
    # Basic cleaning: remove None values and empty strings
    cleaned_recipe = {k: v for k, v in recipe.items() if v is not None and v != "" and v != "Título no encontrado" and v != "Resumen no encontrado" and v != "Ingredientes no encontrados" and v != "Pasos no encontrados" and v != "Tiempo no encontrado" and v != "Porciones no encontradas" and v != "Rating no encontrado"}

    # Add cleaned recipe to the list if it has essential fields
    if all(key in cleaned_recipe for key in ["title", "ingredients", "steps"]):
        cleaned_recipe_data.append(cleaned_recipe)

print(f"Se limpiaron {len(recipe_data) - len(cleaned_recipe_data)} recetas.")
print(f"Número de recetas limpiadas: {len(cleaned_recipe_data)}")

# Display first few cleaned recipes
for i, recipe in enumerate(cleaned_recipe_data[:5]):
    print(f"\nReceta {i+1}:")
    print(f"Título: {recipe.get('title', 'N/A')}")
    print(f"Ingredientes: {recipe.get('ingredients', 'N/A')}")
    print(f"Pasos: {recipe.get('steps', 'N/A')}")
```

```python
from sentence_transformers import SentenceTransformer

# Load a pre-trained model. 'all-MiniLM-L6-v2' is a good general-purpose model.
model = SentenceTransformer('all-MiniLM-L6-v2')

# Prepare the text data for embedding.
# We'll concatenate relevant fields like title, ingredients, and steps.
corpus = []
for recipe in cleaned_recipe_data:
    text = f"Title: {recipe.get('title', '')}\nIngredients: {recipe.get('ingredients', '')}\nSteps: {recipe.get('steps', '')}"
    corpus.append(text)

# Generate embeddings
corpus_embeddings = model.encode(corpus, convert_to_numpy=True)

print(f"Número de recetas en el corpus: {len(corpus)}")
print(f"Forma de las incrustaciones: {corpus_embeddings.shape}")
```

```python
from annoy import AnnoyIndex
import numpy as np

# Define the dimensionality of our embeddings
f = corpus_embeddings.shape[1]

# Create an Annoy index. 'angular' is a common distance metric for embeddings.
t = AnnoyIndex(f, 'angular')

# Add the embeddings to the index
for i, embedding in enumerate(corpus_embeddings):
    t.add_item(i, embedding)

# Build the index with a specified number of trees. More trees provide higher accuracy but take longer to build.
t.build(10) # 10 trees

# Optionally, save the index to disk
# t.save('recipes.ann')

print("Índice Annoy creado y poblado.")
print(f"Número de elementos en el índice: {t.get_n_items()}")
```

```python
# Function to perform RAG search
def search_recipes_rag(query, index, model, recipes_data, k=5):
    """
    Performs a RAG search for recipes based on a query.

    Args:
        query (str): The user's search query.
        index (AnnoyIndex): The Annoy index containing recipe embeddings.
        model (SentenceTransformer): The model used to generate embeddings.
        recipes_data (list): The list of cleaned recipe dictionaries.
        k (int): The number of nearest neighbors to retrieve from the index.

    Returns:
        list: A list of dictionaries containing the retrieved recipe data.
    """
    # 1. Embed the query
    query_embedding = model.encode(query, convert_to_numpy=True)

    # 2. Search the Annoy index for nearest neighbors
    nearest_neighbors_indices = index.get_nns_by_vector(query_embedding, k, include_distances=False)

    # 3. Retrieve the corresponding recipe data
    retrieved_recipes = [recipes_data[i] for i in nearest_neighbors_indices]

    # 4. (Optional) Use a language model to generate a coherent response
    # This part requires a language model and would involve feeding the query and retrieved recipes to the model.
    # For this example, we will just return the retrieved recipes.

    return retrieved_recipes

# Example usage:
# query = "recetas de postres con chocolate"
# retrieved_recipes = search_recipes_rag(query, t, model, cleaned_recipe_data)

# print("\nRecetas encontradas:")
# for recipe in retrieved_recipes:
#     print(f"Título: {recipe.get('title', 'N/A')}")
#     print(f"Ingredientes: {recipe.get('ingredients', 'N/A')}")
#     print("-" * 20)
```

```python
# Example usage:
query = "postres rapidos con pocos ingredientes"
retrieved_recipes = search_recipes_rag(query, t, model, cleaned_recipe_data, k=3)

print(f"Recetas encontradas para la consulta: '{query}'")
for i, recipe in enumerate(retrieved_recipes):
    print(f"\nReceta {i+1}:")
    print(f"Título: {recipe.get('title', 'N/A')}")
    print(f"Ingredientes: {recipe.get('ingredients', 'N/A')}")
    print(f"Pasos: {recipe.get('steps', 'N/A')}")
    print("-" * 20)
```
