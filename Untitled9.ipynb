```python
#librerias
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
#cargar arhivo
file = '/mnt/recipy.html'

with open(file, 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

```python
title = soup.find("meta", {"property": "og:title"})["content"]
print("Titulo: ",title)
```

```python
ingredients_section = soup.find_all("li", class_='mm-recipes-structured-ingredients__list-item')
print("Ingredientes: ")
for ingredient in ingredients_section:
    print(ingredient.text.strip())
```

```python
details_divs = soup.find_all("div", class_="mm-recipes-details__value")
details_text = None

if len(details_divs) > 2:
    details_text = details_divs[2].get_text(strip=True)

print("Tiempo total: ",details_text)
```

```python
rating_section = soup.find("div", id='mm-recipes-review-bar__rating_1-0')
for value in rating_section:
    print("Rating: ", value)
```

```python
details_divs = soup.find_all("div", class_="mm-recipes-details__value")
details_text = None

if len(details_divs) > 3:
    details_text = details_divs[3].get_text(strip=True)

print("Porciones: ",details_text)
```

```python
preparation_section = soup.find_all("div", class_="comp mm-recipes-steps__content mntl-sc-page mntl-block")

if preparation_section:
    print("Preparacion: ")
    for step_group in preparation_section:
        steps = step_group.find_all("p", class_="comp mntl-sc-block mntl-sc-block-html")
        for step in steps:
            print(step.get_text(strip=True))
```

```python
recipes_links = soup.find_all("a")
recipe_urls = []
for link in recipes_links:
    href = link.get('href')
    if "recipe" in href:
        recipe_urls.append(href)
print ("Linked recipes:")
for url in recipe_urls:
    print(url)
print(len(recipe_urls))
```
