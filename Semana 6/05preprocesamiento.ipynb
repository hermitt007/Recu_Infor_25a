## Parte 0: Carga del Corpus

```python
from sklearn.datasets import fetch_20newsgroups

newsgroups = fetch_20newsgroups(subset='all', remove=('headers', 'footers', 'quotes'))
newsgroupsdocs = newsgroups.data
```

## Parte 1: Tokenización

### Actividad
1. Tokeniza los documentos.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab') # Add this line to download the missing resource

from nltk.tokenize import word_tokenize

tokenized_docs = [word_tokenize(doc) for doc in newsgroupsdocs]
print(f"Número de documentos tokenizados: {len(tokenized_docs)}")
print(f"Primer documento tokenizado: {tokenized_docs[0][:20]}...") # Mostrar solo las primeras 20 tokens del primer documento
```

## Parte 2: Normalización

### Actividad
1. Convierte todos los tokens a minúsculas.
2. Elimina puntuación y símbolos no alfabéticos.

```python
# prompt: Convierte todos los tokens a minúsculas.
# Elimina puntuación y símbolos no alfabéticos.

import re

# 1. Convierte todos los tokens a minúsculas.
lower_tokens = [[token.lower() for token in doc] for doc in tokenized_docs]

# 2. Elimina puntuación y símbolos no alfabéticos.
cleaned_tokens = [[re.sub(r'[^a-zA-Z]', '', token) for token in doc] for doc in lower_tokens]

cleaned_tokens = [[token for token in doc if token] for doc in cleaned_tokens]

print(f"Número de documentos normalizados: {len(cleaned_tokens)}")
print(f"Primer documento normalizado: {cleaned_tokens[0][:20]}...") # Mostrar solo las primeras 20 tokens del primer documento
```

## Parte 3: Eliminación de Stopwords

### Actividad
1. Elimina las palabras vacías usando una lista estándar.

```python
# prompt: Elimina las palabras vacías usando una lista estándar.

nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
filtered_tokens = [[token for token in doc if token not in stop_words] for doc in cleaned_tokens]

print(f"Número de documentos después de eliminar stopwords: {len(filtered_tokens)}")
print(f"Primer documento después de eliminar stopwords: {filtered_tokens[0][:20]}...")
```

## Parte 4: Stemming o Lematización

### Actividad
1. Aplica stemming.
2. Aplica lematización.
3. Compara ambas técnicas.

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet, stopwords
from nltk.tokenize import word_tokenize

# Descargar recursos necesarios
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')

# ---------------------------
# Simulamos documentos (ejemplo)
# ---------------------------
sample_docs = [
    "Running is better than walking when you're in a hurry.",
    "Cats are chasing the mice quickly and efficiently.",
]

# Tokenizar los documentos
tokenized_docs = [word_tokenize(doc.lower()) for doc in sample_docs]

# Eliminar stopwords
stop_words = set(stopwords.words('english'))
filtered_tokens = [
    [token for token in doc if token.isalpha() and token not in stop_words]
    for doc in tokenized_docs
]

# ---------------------------
# 1. Aplicar Stemming
# ---------------------------
porter = PorterStemmer()
stemmed_tokens = [[porter.stem(token) for token in doc] for doc in filtered_tokens]

print(f"Número de documentos después de stemming: {len(stemmed_tokens)}")
print(f"Primer documento después de stemming: {stemmed_tokens[0][:20]}...")

# ---------------------------
# 2. Aplicar Lematización
# ---------------------------
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(word):
    """Mapea etiquetas POS a las usadas por WordNetLemmatizer"""
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {
        "J": wordnet.ADJ,
        "N": wordnet.NOUN,
        "V": wordnet.VERB,
        "R": wordnet.ADV,
    }
    return tag_dict.get(tag, wordnet.NOUN)

lemmatized_tokens = [
    [lemmatizer.lemmatize(token, get_wordnet_pos(token)) for token in doc]
    for doc in filtered_tokens
]

print(f"Número de documentos después de lematización: {len(lemmatized_tokens)}")
print(f"Primer documento después de lematización: {lemmatized_tokens[0][:20]}...")

# ---------------------------
# 3. Comparación palabra por palabra
# ---------------------------
def comparar_palabra(palabra):
    stemmed = porter.stem(palabra)
    lemmatized = lemmatizer.lemmatize(palabra, get_wordnet_pos(palabra))
    print(f"\nPalabra original: '{palabra}'")
    print(f"Stemming: {stemmed}")
    print(f"Lematización: {lemmatized}")

comparar_palabra("running")
comparar_palabra("better")
comparar_palabra("houses")
```
